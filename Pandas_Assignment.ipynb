```python
import pandas as pd
import numpy as np
```

1. Create DataFrame and Add Product Column

```python
df = pd.DataFrame(np.random.randint(1, 10, (5, 3)),
                  columns=['A', 'B', 'C'])

print("Original DataFrame:")
print(df)

df['Product'] = df['A'] * df['B']

print("Updated DataFrame:")
print(df)
```

2. Row-wise and Column-wise Sum

```python
df = pd.DataFrame(np.random.randint(1, 10, (4, 3)),
                  columns=['A', 'B', 'C'])

print("DataFrame:")
print(df)

print("Row-wise Sum:")
print(df.sum(axis=1))

print("Column-wise Sum:")
print(df.sum(axis=0))
```

3. Handling NaN Values Using Mean

```python
df = pd.DataFrame(np.random.randint(1, 20, (5, 3)),
                  columns=['A', 'B', 'C'])

df.iloc[1, 0] = np.nan
df.iloc[2, 1] = np.nan
df.iloc[3, 2] = np.nan

print("Original DataFrame:")
print(df)

df = df.fillna(df.mean())

print("DataFrame after filling NaN:")
print(df)
```

4. Drop Rows Containing NaN Values

```python
df = pd.DataFrame(np.random.randint(1, 20, (6, 4)),
                  columns=['A', 'B', 'C', 'D'])

df.iloc[1, 0] = np.nan
df.iloc[3, 2] = np.nan
df.iloc[4, 1] = np.nan

print("Original DataFrame:")
print(df)

df = df.dropna()

print("DataFrame after dropping NaN rows:")
print(df)
```

5. Data Aggregation – Sum and Mean

```python
df = pd.DataFrame({
    'Category': np.random.choice(['A', 'B', 'C'], 10),
    'Value': np.random.randint(1, 100, 10)
})

print("Original DataFrame:")
print(df)

result = df.groupby('Category')['Value'].agg(['sum', 'mean'])

print("Grouped DataFrame:")
print(result)
```

6. Group by Category and Calculate Total Sales

```python
df = pd.DataFrame({
    'Product': ['Pen', 'Book', 'Pencil', 'Bag', 'Bottle', 'Box'],
    'Category': ['A', 'B', 'A', 'C', 'B', 'C'],
    'Sales': [100, 200, 150, 300, 250, 180]
})

print("Original DataFrame:")
print(df)

result = df.groupby('Category')['Sales'].sum()

print("Total Sales by Category:")
print(result)
```

7. Merging DataFrames

```python
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Rahul', 'Anu', 'Kiran']
})

df2 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Marks': [85, 90, 88]
})

print("First DataFrame:")
print(df1)

print("Second DataFrame:")
print(df2)

result = pd.merge(df1, df2, on='ID')

print("Merged DataFrame:")
print(result)
```

8. MultiIndex DataFrame

```python
arrays = [
    ['A', 'A', 'B', 'B'],
    ['One', 'Two', 'One', 'Two']
]

index = pd.MultiIndex.from_arrays(arrays,
                                  names=['Category', 'Type'])

df = pd.DataFrame({'Value': [10, 20, 30, 40]},
                  index=index)

print("MultiIndex DataFrame:")
print(df)

print("Accessing Category A:")
print(df.loc['A'])

print("Accessing A and One:")
print(df.loc[('A', 'One')])

print("Slicing DataFrame:")
print(df.loc['A':'B'])
```

9. MultiIndex Grouping and Sum

```python
df = pd.DataFrame({
    'Category': ['A', 'A', 'A', 'B', 'B', 'B'],
    'SubCategory': ['X', 'Y', 'X', 'X', 'Y', 'Y'],
    'Value': np.random.randint(1, 100, 6)
})

df = df.set_index(['Category', 'SubCategory'])

print("MultiIndex DataFrame:")
print(df)

result = df.groupby(level=['Category', 'SubCategory']).sum()

print("Sum of Values:")

print(result)
```

10. Applying Functions – Double the Values

```python
df = pd.DataFrame(np.random.randint(1, 10, (5, 3)),
                  columns=['A', 'B', 'C'])

print("Original DataFrame:")
print(df)

result = df.apply(lambda x: x * 2)

print("DataFrame after doubling:")
print(result)
```

11. Apply Lambda Function to Create a New Column

```python
df = pd.DataFrame(np.random.randint(1, 10, (6, 3)),
                  columns=['A', 'B', 'C'])

print("Original DataFrame:")
print(df)

df['Total'] = df.apply(lambda x: x['A'] + x['B'] + x['C'], axis=1)

print("Updated DataFrame:")
print(df)
```

12. Working with Text Data – Uppercase

```python
s = pd.Series(['apple', 'banana', 'mango', 'orange', 'grapes'])

print("Original Series:")
print(s)

result = s.str.upper()

print("Uppercase Series:")
print(result)
```

13. Extract First Three Characters

```python
s = pd.Series(['apple', 'banana', 'mango', 'orange', 'grapes'])

print("Original Series:")
print(s)

result = s.str[:3]

print("First Three Characters:")
print(result)
```

1. Counting Words in a File

```python
def count_words():
    with open("document.txt", "r") as file:
        content = file.read()
        words = content.split()
        return len(words)
print("Number of words:", count_words())
```

2. Finding and Replacing Text

```python
def find_replace(old_word, new_word):
    with open("data.txt", "r") as file:
        content = file.read()
    content = content.replace(old_word, new_word)
    with open("data.txt", "w") as file:
        file.write(content)
    print("Text replaced successfully.")
find_replace("Hello", "Hi")
```

3. Reading a File in Reverse

```python
def reverse_lines():
    with open("reverse.txt", "r") as file:
        lines = file.readlines()

    for line in lines:
        print(line.rstrip()[::-1])

reverse_lines()
```

4. Counting Lines, Words, and Characters

```python
def count_stats():
    with open("stats.txt", "r") as file:
        content = file.read()

    lines = len(content.splitlines())
    words = len(content.split())
    characters = len(content)

    return lines, words, characters

l, w, c = count_stats()

print("Number of lines:", l)
print("Number of words:", w)
print("Number of characters:", c)
```

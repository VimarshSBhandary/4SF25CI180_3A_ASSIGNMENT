```python
import numpy as np
```

1. Array Creation and Manipulation

```python
array = np.random.randint(1, 21, (5, 5))
print("Original Array:")
print(array)
array[:, 2] = 1
print("Modified Array:")
print(array)
```

2. Replace Diagonal Elements with 0

```python
array= np.arange(1, 17).reshape(4, 4)
print("Original Array:")
print(array)
np.fill_diagonal(array, 0)
print("Modified Array:")
print(array)
```

3. Extract Elements on the Border

```python
array = np.random.randint(1, 21, (5, 5))

print("Original Array:")
print(array)

border = np.concatenate((
    array[0, :],
    array[1:-1, -1],
    array[-1, ::-1],
    array[-2:0:-1, 0]
))

print("Border Elements:")
print(border)
```

4. Row-wise and Column-wise Sum

```python
array = np.arange(1, 17).reshape(4, 4)

print("Original Array:")
print(array)

row_sum = np.sum(array, axis=1)
column_sum = np.sum(array, axis=0)

print("Row-wise Sum:")
print(row_sum)

print("Column-wise Sum:")
print(column_sum)
```

5. Reshape the Array

```python
array= np.arange(1, 10).reshape(3, 3)

print("Original Array:")
print(array)

array1 = array.reshape(1, 9)
print("Array of shape (1, 9):")
print(array1)

array2 = array1.reshape(9, 1)
print("Array of shape (9, 1):")
print(array2)
```

6. Structured Array and Sorting

```python
dtype = [
    ('name', 'U20'),
    ('age', 'i4'),
    ('weight', 'f4')
]

array= np.array([
    ('Rahul', 22, 65.5),
    ('Anu', 20, 52.5),
    ('Kiran', 24, 70.2),
    ('Priya', 21, 55.8)
], dtype=dtype)

print("Original Array:")
print(array)

sorted_arr = np.sort(array, order='age')

print("Sorted Array by Age:")
print(sorted_arr)
```
